```python
# ruff: noqa: F401
import os

import requests
from deepagents import create_deep_agent
from deepagents.backends import FilesystemBackend
from dotenv import load_dotenv
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from rich.console import Console
from rich.markdown import Markdown
from slide_kit import build_slide

load_dotenv(override=True)
console = Console()
```

```python
@tool
def serper_search(query: str) -> str:
    """Search Google using the Serper API."""
    url = "https://google.serper.dev/search"
    headers = {
        "X-API-KEY": os.environ.get("SERPER_API_KEY", ""),
        "Content-Type": "application/json",
    }
    response = requests.post(url, headers=headers, json={"q": query})
    return response.text
```

```python
sandbox = os.path.abspath("sandbox")
os.makedirs(sandbox, exist_ok=True)
```

### Create Agent


```python
model = ChatOpenAI(model="openai/gpt-5.4-mini")
researcher = create_deep_agent(
    model=model,
    tools=[serper_search],
    system_prompt="You're a research analyst. Plan your work with todo tool, reseach with the search tool,"
    "and report your findings as a tidy markdown briefing to a file",
    backend=FilesystemBackend(root_dir=sandbox, virtual_mode=True),
)
```

```python
researcher
```

```python
brief = """
Our company is planning to move its sales fleet to electric vehicles.
Research the public EV charging landscape in US and find out roughly how many public charging points are there.
Pick out two major charging networks that we can rely on.
Write one page markdown briefing, with a heading and a short section for each, to the file charging.md.
"""

model = ChatOpenAI(model="gpt-4o-mini")

researcher = create_deep_agent(
    model=model,
    tools=[serper_search],
    system_prompt=(
        "You're a research analyst. Plan your work with todo tool, "
        "research with the search tool, and report your findings as a "
        "tidy markdown briefing to a file."
    ),
    backend=FilesystemBackend(root_dir=sandbox, virtual_mode=True),
)

result = researcher.invoke({"messages": [{"role": "user", "content": brief}]})
console.print(Markdown(result["messages"][-1].content))
```

```python
tools_called = [tc["name"] for m in result["messages"] for tc in getattr(m, "tool_calls", []) or []]
print("Tools called in order:")
print(tools_called)
```

```python
for m in result["messages"]:
    tool_called = getattr(m, "tool_calls", [])
    for tc in tool_called:
        print(tc["name"])
```

### Sub-agents


```python
research_ev_instructions = """
You research one electric vehicle using the search tool and return 3 concise facts that a buyer would be inteersted in,
For example, price, mileage and charging.
"""

overall_instructions = """
You write comparison briefings for a company choosing electric vehicles for its own sales fleet.
For each vehicle, delegate the research to the vehicle-researcher sub-agent,
the write down the result in a markdown comparison file, ending with a clear recommended model.
"""

researcher_subagent = {
    "name": "vehicle-researcher",
    "description": "Research a single electric vehicle and returns the facts about it",
    "system_prompt": research_ev_instructions,
}


lead_researcher = create_deep_agent(
    model=ChatOpenAI(model="gpt-5.4-mini"),
    tools=[serper_search],
    system_prompt=overall_instructions,
    subagents=[researcher_subagent],
    backend=FilesystemBackend("sandbox", virtual_mode=True),
)
```

```python
lead_researcher
```

```python
mission = """
Compare between Tesla Model Y and Ford Mustang Mach-E as candidates for 100 car sales fleet.
Research each vehicle and write a comparison with a final recommendation in a markdown file fleet.md
"""

result = lead_researcher.invoke({"messages": [{"role": "user", "content": mission}]})

console.print(Markdown(result["messages"][-1].content))
```

```python
tools_used = [tu["name"] for m in result["messages"] for tu in getattr(m, "tool_calls", []) or []]
print(tools_used)
```

### Using customized skills (SKILL.md)


```python
sandbox = os.path.abspath("sandbox")
os.makedirs(sandbox, exist_ok=True)


# Define custom tool to create powerpoint sldies
@tool
def create_slide(title: str, key_points: list[str], recommendation: str) -> str:
    """Create a one slider powerpoint presentation in the Voltway Research house style and save as fleet.pptx"""
    build_slide(title, key_points, recommendation, os.path.join(sandbox, "fleet.pptx"))
    return "Saved the file to /fleet.pptx"


# Define subagent
slide_maker = {
    "name": "slide-maker",
    "description": "Subagent to create a one-slider powerpoint deck from recommendations",
    "system_prompt": (
        "Read the fleet-slide skill and use it to create a single-slide "
        "PowerPoint. You must call create_slide to save fleet.pptx. "
        "Only report completion after the tool succeeds."
    ),
    "tools": [create_slide],
    "skills": ["/skills/"],
}

presenter = create_deep_agent(
    model="gpt-5.4-mini",
    subagents=[slide_maker],
    system_prompt="Prepare a reaseach presentation by delegating to the slide-maker sub-agent",
    backend=FilesystemBackend(sandbox, virtual_mode=True),
)
```

```python
result = presenter.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": (
                    "Read /fleet.md and delegate to slide-maker to create "
                    "a single-slide PowerPoint saved as /fleet.pptx."
                ),
            }
        ]
    }
)

console.print(Markdown(result["messages"][-1].content))
```

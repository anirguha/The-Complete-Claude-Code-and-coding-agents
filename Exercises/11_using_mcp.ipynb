## Using MCP (Model Context Protocol)


```python
from agents.mcp import MCPServerStreamableHttp
from rich.console import Console
from rich.markdown import Markdown

from agents import Agent, Runner
```

```python
console = Console()
OPENAI_MODEL_NAME = "gpt-4o-mini"
```

```python
task = """
In the new SandboxAgents feature in the OpenAI Agents SDK as of May 2026, what is the role of the Manifest object?
Always be accurate. If you don't know the answer, say so.
"""
```

#### Running without MCP on an old model


```python
agent = Agent(name="Expert", instructions="Answer the question", model=OPENAI_MODEL_NAME)
```

```python
result = await Runner.run(agent, task)
result.final_output
```

### Running with MCP on an old model


```python
params = {"url": "https://mcp.context7.com/mcp", "timeout": 60}

async with MCPServerStreamableHttp(name="Context7", params=params) as server:
    agent = Agent(
        name="Expert",
        instructions="Use Context7 to answer the question",
        mcp_servers=[server],
        model="gpt-4o-mini",
    )
    result = await Runner.run(agent, task)
```

```python
console.print(Markdown(result.final_output))
```

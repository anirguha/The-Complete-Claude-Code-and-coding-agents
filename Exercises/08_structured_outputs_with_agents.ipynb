# Use of Pydantic Package for creating structured outputs for agents


```python
from agents import Agent, Runner  # noqa: I001
from pydantic import BaseModel, Field
from rich.console import Console
from rich.markdown import Markdown
from rich.syntax import Syntax

console = Console()
OPEANAI_MODEL_NAME = "gpt-5.4-nano"
```

## Create an object with Pydantic BaseModel


```python
class Email_review(BaseModel):
    is_professional: bool = Field(description="Checks if the email tone is professional")
    num_lines: int = Field(description="Captures the number of lines in the email")
    has_placeholder: bool = Field(description="Checks if the email contains any placeheolder")
```

```python
email_review = Email_review
```

```python
email_review = Email_review(is_professional=True, num_lines=5, has_placeholder=False)
```

```python
console.print(Syntax(code=email_review.model_dump_json(), lexer="json"))
```

```python
instructions = """
Review the potential sales emails
"""

email = """
Hi [first_name],

I'm hitting you up to see if you'd like to buy our product. It's really great. You'll miss out if you don't buy it.

Laters.

AG
"""
```

#### Create Agent


```python
email_reviewer = Agent(
    name="Email Reviewer",
    instructions=instructions,
    output_type=Email_review,
    model=OPEANAI_MODEL_NAME,
)
```

#### Run the Agent


```python
result = await Runner.run(email_reviewer, input=email)
```

```python
console.print(Syntax(result.final_output.model_dump_json(), "json"))
```

```python
result.final_output.is_professional
```

```python
console.print(Markdown(result.final_output.model_dump_json(), "json"))
```

```python
emails2 = """
Dear Mr. Guha,

I'd like to bring to your notice the wonderful product our company has 
launched recently. As a valued customer I would like you to try the product 
and provide your invaluable comments.

Laters.

AG
"""
```

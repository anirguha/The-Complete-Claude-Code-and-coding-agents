```python
# ruff: noqa: F401
import os

from agents.mcp import MCPServerStdio, MCPServerStreamableHttp
from dotenv import load_dotenv
from IPython.display import Image, display
from rich.console import Console
from rich.markdown import Markdown

from agents import Agent, RunHooks, Runner, trace

load_dotenv(override=True)
console = Console()
```

```python
class ActivityHooks(RunHooks):
    def __init__(self):
        self.turn_count = 0

    async def on_agent_start(self, context, agent):
        print(f"\nAgent started: {agent.name}")

    async def on_llm_start(self, context, agent, system_prompt, input_items):
        self.turn_count += 1
        print(f"\n--- Turn {self.turn_count}: {agent.name} calling model ---")

    async def on_llm_end(self, context, agent, response):
        print(f"[Turn {self.turn_count}] Model response:")
        for item in response.output:
            print(f"  {item.type}: {item}")

    async def on_tool_start(self, context, agent, tool):
        print(f"[Turn {self.turn_count}] Starting tool: {tool.name}")

    async def on_tool_end(self, context, agent, tool, result):
        print(f"[Turn {self.turn_count}] Tool finished: {tool.name}")
        print(f"  Result: {result}")
```

### Add Fetch MCP server (Python based)


```python
fetch_params = {"command": "uvx", "args": ["mcp-server-fetch"]}

async with MCPServerStdio(params=fetch_params, client_session_timeout_seconds=60) as server:
    fetch_tools = await server.list_tools()

print(fetch_tools)
```

### Add Playwright MCP server (JS based)


```python
playwright_params = {"command": "npx", "args": ["@playwright/mcp@latest"]}

async with MCPServerStdio(params=playwright_params, client_session_timeout_seconds=60) as server:
    playwright_tools = await server.list_tools()

print(len(playwright_tools))
print(*playwright_tools, sep="\n")
```

### Add sandbox tool (JS based)


```python
sandbox_path = os.path.abspath(os.path.join(os.getcwd(), "sandbox"))
os.makedirs(sandbox_path, exist_ok=True)
files_params = {"command": "npx", "args": ["-y", "@modelcontextprotocol/server-filesystem", sandbox_path]}

async with MCPServerStdio(params=files_params, client_session_timeout_seconds=60) as server:
    file_tools = await server.list_tools()

print(len(file_tools))
print(*file_tools, sep="\n")
```

## Use MCP Servers in agents


### Using Standard Input/ Output Transfer (f0r local MCP Servers)


```python
INSTRUCTIONS = """
You are are agent that browses the internet to accomplish tasks.
Accept cookies and navigate pop-ups as needed.
if one website isn't useful try another.
Be persistent until the task is accomplished.
Try different options and sites as needed.
When you need to srite files use the files in sandbox folder only.
"""

TASK = "Find the recipe for Banoffee Pie and summarize it in banoffee.md"

async with MCPServerStdio(files_params, client_session_timeout_seconds=60) as mcp_server_files:
    async with MCPServerStdio(playwright_params, client_session_timeout_seconds=60) as mcp_server_browser:
        agent = Agent(
            name="Investigator",
            instructions=INSTRUCTIONS,
            model="gpt-5.4-nano",
            mcp_servers=[mcp_server_files, mcp_server_browser],
        )
        hooks = ActivityHooks()
        with trace("investigate"):
            result = await Runner.run(agent, TASK, max_turns=20, hooks=ActivityHooks())
            console.print(Markdown(result.final_output))
            print(f"\nTotal turns: {hooks.turn_count}")
```

### Using Streamable HTTP Transfer (for remote MCP server)


```python
question = """
In the SandboxAgents feature added to OpenAI SDK in 2026, what is the use of Manifest object?
Provide an accurate answer, if you are aware, otherwise please mention that you are not aware.
"""

agent = Agent(name="Expert", instructions="Answer the question given as the prompt", model="gpt-5.4-nano")

result = await Runner.run(starting_agent=agent, input=question)
console.print(Markdown(result.final_output))
```

```python
params = {"url": "https://mcp.context7.com/mcp", "timeout": 60}

async with MCPServerStreamableHttp(params, name="context7") as server:
    agent = Agent(
        name="Expert",
        instructions="Use context7 to answer the question with an example",
        model="gpt-5.4-nano",
        mcp_servers=[server],
    )
    result = await Runner.run(agent, question)

console.print(Markdown(result.final_output))
```

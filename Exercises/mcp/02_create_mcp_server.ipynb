```python
# ruff: noqa: F401
from agents.mcp import MCPServerStdio
from backend.accounts import Account
from dotenv import load_dotenv
from IPython.display import Image, display
from rich.console import Console
from rich.markdown import Markdown

from agents import Agent, Runner, trace

load_dotenv(override=True)
console = Console()
```

### Set up custom MCP Server


```python
params = {"command": "uv", "args": ["run", "-m", "backend.accounts_server"]}
async with MCPServerStdio(params, client_session_timeout_seconds=60) as server:
    tools_list = await server.list_tools()

print(*tools_list, sep="\n")
```

### Use the MCP server in an agent


```python
INSTRUCTIONS = """
You are an account manager who can answer questions or perform a task about an account from a client.
"""

question = """
My name is AG. Show me my account holdings.
"""

TASK = """
I want to buy 3 shared of Apple with ticker sign AAPL. Show the current holdings at the end.
"""

async with MCPServerStdio(params=params, client_session_timeout_seconds=60) as server:
    account_manager = Agent("account-manager", mcp_servers=[server], instructions=INSTRUCTIONS, model="gpt-5.4-nano")
    with trace("account_manager"):
        response = await Runner.run(account_manager, question)
    display(Markdown(response.final_output))
```
